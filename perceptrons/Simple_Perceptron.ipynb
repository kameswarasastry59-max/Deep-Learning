## Creating a simple Perceptron to perform ADD Operation 

```python
from percetron import Perceptron
import numpy as np 

# 1. Define the AND gate training data
# X represents the inputs (feature 1, feature 2)
X_train = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# y represents the correct answers (labels)
y_train = np.array([0, 0, 0, 1])

# 2. Create the Perceptron object
# It will use the default learning_rate=0.1 and epochs=10 from our __init__
model = Perceptron()


# 3. Train the model
model.fit(X_train, y_train)

# 4. Let's see what it learned!
print("Learned Weights:", model.weights)
print("Learned Bias:", model.bias)

# 5. Make predictions
predictions = model.predict(X_train)
print("Predictions:", predictions)
print("Real Answers:", y_train)
```

## Iris Classification 

```python
from percetron import Perceptron
import numpy as np 
from sklearn.datasets import load_iris

# 1. Import and define the training data
# Load the official dataset
iris = load_iris()

# The first 100 rows contain 50 Setosas (labeled 0) and 50 Versicolors (labeled 1)
# We will grab those 100 rows, and only take columns 2 and 3 (Petal Length and Width)
X_train = iris.data[:100, 2:4] 

# Get the answers for those 100 flowers
y_original = iris.target[:100]

# Let's flip the labels to match our old example: 1 = Setosa, 0 = Not Setosa
y_train = np.where(y_original == 0, 1, 0)


# 2. Create the Perceptron object
# We might need a few more passes (epochs) since we have 100 flowers now!
model = Perceptron(learning_rate=0.1, epoches=20)


# 3. Train the model
model.fit(X_train, y_train)


# 4. Let's see what it learned!
print("Learned Weights:", model.weights)
print("Learned Bias:", model.bias)


# 5. Make predictions
predictions = model.predict(X_train)

# Instead of printing 100 rows, let's calculate the accuracy percentage!
correct_predictions = np.sum(predictions == y_train)
total_predictions = len(y_train)
accuracy = (correct_predictions / total_predictions) * 100

print(f"\nModel Accuracy: {accuracy}%")
print(f"Correctly guessed {correct_predictions} out of {total_predictions} flowers.")
```
