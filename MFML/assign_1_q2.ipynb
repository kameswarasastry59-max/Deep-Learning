```python
# Helper code to print the matrix with alignment 

def print_matrix_nicely(mat, name="Matrix",  precision=8, limit=None):
    """Print a matrix with aligned values."""
    rows = len(mat)
    cols = len(mat[0]) if rows > 0 else 0
    
    # Prints the header with matrix name and dimension
    print(f"{name} \n (Shape : {rows}x{cols}) ---")

    # Optional limit to restrict the num of rows displayed.
    if not limit:
        limit=len(mat)
    
    # Formats and align each cell
    for row in mat[:limit]:
        values = [f"{float(value):12.{precision}f}" for value in row]
        print(" [" + " ".join(values) + " ]")
```

##### Q) 2.1 Python code to generate the dataset X = [f1, f2, f3, f4, f5, f6].

```python
import math
import random

def generate_dataset(n_samples=500):
    """
    Generates a dataset X of size n_samples x 6 using pure Python.
    The first four features are from standard normal distriution. The 5th and 6th are linear combinations.
    """
    random.seed(42)
    X = []
    
    for _ in range(n_samples):
        # features f1-f4 are generated as standard normal using random.gauss 
        f1 = random.gauss(0, 1)  
        f2 = random.gauss(0, 1)
        f3 = random.gauss(0, 1)
        f4 = random.gauss(0, 1)
        f5 = 2 * f1 + 3 * f2
        f6 = f3 - 2 * f4
        
        X.append([f1, f2, f3, f4, f5, f6])
        
    return X
# 1. Generate Dataset
X = generate_dataset()
print(f"Task 1: Dataset X generated with {len(X)} rows and {len(X[0])} columns.")
print("\n")
# helper function called to show the generated dataset
print_matrix_nicely(X, name="Generated Dataset Preview: ", limit=10)
```

#### Q) 2.2 Python code which computes the rank of X and display the output for the Generated dataset X

```python
def compute_rank(matrix):
    """
    Computes the rank of a matrix using Gaussian elimination from scratch.
    
    Algorithm

    1. Set the rank to 0 and loop through each column of the matrix.
    2. For the current column, search downward from the current rank row for a non-zero pivot. Values smaller than 1e -10 are treated as zero.
    3. If no pivot is found, skip the column and move to the next one.
    4. If a pivot is found, swap its row with the current rank row.
    5. Use the pivot row to eliminate all elements below the pivot by subtracting a suitable multiple of the pivot row.
    6. Increment the rank by 1 after successfully processing a pivot.
    7. Continue until all columns have been checked.
    8. The final value of the rank counter is the rank of the matrix.

    """
    # Create a deep copy to avoid modifying the original dataset
    mat = [row[:] for row in matrix]
    rows = len(mat)
    cols = len(mat[0])
    rank = 0
    
    for c in range(cols):
        pivot = rank
        while pivot < rows and abs(mat[pivot][c]) < 1e-10:
            pivot += 1
            
        if pivot < rows:
            # Swap rows
            mat[rank], mat[pivot] = mat[pivot], mat[rank]
            # Eliminate below
            for r in range(rank + 1, rows):
                factor = mat[r][c] / mat[rank][c]
                for k in range(c, cols):
                    mat[r][k] -= factor * mat[rank][k]
            rank += 1
            
    return rank

# 2. Compute Rank
rank_X = compute_rank(X)
print(f"Computed Rank of the Generated Dataset X: {rank_X}")
```

#### Q) 2.3 (a) Compute Covariance Matrix

```python
def transpose(matrix: list) -> list:
    """Returns the transpose of a matrix."""
    return [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]

def mat_mult(A: list, B: list) -> list:
    """Multiplies two matrices A and B."""
    B_T = transpose(B)
    return [[sum(a * b for a, b in zip(row_a, col_b)) for col_b in B_T] for row_a in A]

def compute_covariance(matrix: list) -> list:
    """
    Computes the sample covariance-style matrix C = (1/n) X^T X.

    Algorithm: 

    1. Determine the total number of rows in the input matrix and store it as n.
    2. Compute the transpose of the input matrix using a helper function.
    3. Multiply the transposed matrix by the original matrix.
    4. Divide every element of the resulting matrix by n to scale the values.
    5. Return the resulting matrix as the covariance matrix.
    """
    n = len(matrix)
    X_T = transpose(matrix)
    X_T_X = mat_mult(X_T, matrix)
    
    # Divide by n
    return [[val / n for val in row] for row in X_T_X]

# 3(a) Compute Covariance Matrix
C = compute_covariance(X)
# helper function used to print the Covariance Matrix
print_matrix_nicely(C, name="Covariance_Matrix")
```

#### Q) 2.3 (b) Using Power Method to get Largest Eigenvalue and corresponding Eigenvector 

```python
def vector_norm(v):
    """Computes the Euclidean norm of a vector."""
    return math.sqrt(sum(x**2 for x in v))

def mat_vec_mult(M, v):
    """Multiplies a matrix by a vector."""
    return [sum(m * x for m, x in zip(row, v)) for row in M]

def vec_dot(v1, v2):
    """Computes the dot product of two vectors."""
    return sum(x * y for x, y in zip(v1, v2))

def power_method(matrix, num_iterations= 1000, tol = 1e-9):
    """
    Approximates the largest eigenvalue and its corresponding eigenvector using the Power Method.
    
    Algorithm:
    1. Create an initial guess vector filled with 1s and normalize it so its length is exactly 1.
    2. Start a loop that runs up to the maximum number of iterations.
    3. Multiply the input matrix by the current vector to produce the next vector guess.
    4. Normalize this new vector by dividing it by its Euclidean length.
    5. Estimate the new eigenvalue by computing the dot product of the new vector and the matrix-multiplied new vector.
    6. Check the difference between this new eigenvalue and the previous one. If it is smaller than the tolerance threshold,
       the algorithm has converged.
    7. If converged, return the eigenvalue and eigenvector. If not, update the stored vector and eigenvalue and repeat the loop.
    """
    n = len(matrix)
     # Start with a vector of ones
    v = [1.0] * n

    # Normalize the initial vector
    norm_v = vector_norm(v)
    v = [x / norm_v for x in v]

    eigenvalue = 0.0

    for _ in range(num_iterations):
        
        # Get the next vector
        v_next = mat_vec_mult(matrix, v)

        # Normalize the new vector
        norm_v_next = vector_norm(v_next)
        v_next = [x / norm_v_next for x in v_next]

        # Estimate the eigenvalue
        M_v_next = mat_vec_mult(matrix, v_next)
        eigenvalue_next = vec_dot(v_next, M_v_next)

        # Check for convergence
        if abs(eigenvalue_next - eigenvalue) < tol:
            return eigenvalue_next, v_next

        # Update for the next iteration
        v = v_next
        eigenvalue = eigenvalue_next

    return eigenvalue, v

# 3(b) Implement Power Method
lambda_1, v_1 = power_method(C)
print("Power Method Largest Eigenvalue (λ1):")
print(round(lambda_1, 8))
print("Eigenvector (v1):")
print([round(x, 8) for x in v_1])
```

#### Q) 2.3 (c) Apply matrix deflation iteratively to extract the top k eigenvalues and eigenvectors


* `Matrix Deflation`: Remove the influence of previously found dominant eigenvectors by updating the matrix using $C - \sum_{j=1}^{k-1} v_j v_j^T C$.
* `Sequential Discovery`:Execute the created power method on this newly deflated matrix to extract the next largest eigenvalue $\lambda_k$ and its corresponding eigenvector $v_k$.

```python
def outer_product(v1, v2):
    """Computes the outer product of two vectors."""
    return [[x * y for y in v2] for x in v1]

def mat_sub(A, B):
    """Subtracts matrix B from matrix A."""
    return [[A[i][j] - B[i][j] for j in range(len(A[0]))] for i in range(len(A))]

def mat_add(A, B):
    """Adds matrix A and matrix B."""
    return [[A[i][j] + B[i][j] for j in range(len(A[0]))] for i in range(len(A))]

def find_all_eigenvalues_power_method(C_orig, k):
    """
    Applies matrix deflation iteratively to extract the top k eigenvalues and eigenvectors.
    
    1. Initialize empty lists to store the discovered eigenvalues and eigenvectors.
    2. Create a working copy of the original input matrix to modify as we go.
    3. Start a loop that runs exactly k times to extract the top k eigen-pairs.
    4. In each step, apply the standard power method on the current working matrix to find its 
       dominant eigenvalue and eigenvector, and save them.
    5. Construct a deflation matrix by calculating the outer product of every previously found 
       eigenvector with itself, multiplying that by the original matrix, and adding them all together.
    6. Update the working matrix by subtracting this total deflation term from the original matrix, 
       effectively zeroing out the dominant eigenvalues we already found.
    7. Once the loop finishes, return the final lists of eigenvalues and eigenvectors.
    """
    n = len(C_orig)
    # intialize eigenvalues and eignevectors 
    eigenvalues = []
    eigenvectors = []
    # Deep copy (working copy)
    C_current = [row[:] for row in C_orig]
    
    for _ in range(k):
        # intial eigenval and eigenvector using standered power method 
        val, vec = power_method(C_current)
        eigenvalues.append(val)
        eigenvectors.append(vec)
        
        # Deflate using the formula: C_current = C_orig - sum(v_j v_j^T C_orig)
        deflation_term = [[0.0] * n for _ in range(n)]
        for j in range(len(eigenvectors)):
            v_j = eigenvectors[j]
            vj_vjT = outer_product(v_j, v_j)
            vj_vjT_C = mat_mult(vj_vjT, C_orig)
            deflation_term = mat_add(deflation_term, vj_vjT_C)
            
        C_current = mat_sub(C_orig, deflation_term)
        
    return eigenvalues, eigenvectors

# 3(c) Find successive eigenvalues. Rank is 4, so we search for top 4.
vals_pm, vecs_pm = find_all_eigenvalues_power_method(C, k=4)

print("Eigenvalues extracted via Deflation Power Method:")
for idx, val in enumerate(vals_pm):
    print(f"λ{idx+1} = {val:.8f}")
```

#### Q) 2.3 (d) Use the NumPy Built to get the eignevalues and comare that with eigenvalues created using the Power Method

```python
import numpy as np

def get_actual_eigenvalues_numpy(matrix):
    """
    Computes exact eigenvalues using NumPy's built-in solver for symmetric matrices.
    
    1. Converts the pure Python 2D list into a NumPy array.
    2. Passes the array to np.linalg.eigh(), which calculates the exact eigen-pairs.
    3. Sorts the resulting eigenvalues in descending order to properly align with the Power Method's sequential extraction.
    4 .Returns the sorted values as a standard Python list.
    """
    C_np = np.array(matrix)
    
    # np.linalg.eigh is explicitly designed for symmetric matrices like C
    actual_vals, actual_vecs = np.linalg.eigh(C_np)
    
    # Sort in descending order to match the deflation sequence
    sorted_vals = np.sort(actual_vals)[::-1]
    return sorted_vals.tolist()

# 3(d) Find all using NumPy built-in function and compare
# 3(d) Find all using NumPy built-in function and compare
actual_vals_np = get_actual_eigenvalues_numpy(C)

print("Task 3(d): Built-in Eigenvalues vs Power Method")

comparison = []

for idx, val in enumerate(actual_vals_np[:4]):
    power_val = vals_pm[idx]
    difference = abs(val - power_val)
    comparison.append([val, power_val, difference])

print("\nBuilt-in Eigenvalue | Power Method | Difference")
print(np.array(comparison))
```

```python
def count_iterations_for_accuracy(matrix: list, true_val: float, target_acc: float = 1e-7) -> int:
    """
    Counts the iterations required for the Power Method to reach a specific accuracy.

    1. Initialize a guess vector with all elements set to 1 and normalize it to have unit length.
    2. Set the iteration counter to 0.
    3. Multiply the input matrix by the current vector to obtain the next vector.
    4. Normalize the new vector using its Euclidean norm.
    5. Estimate the eigenvalue using the Rayleigh quotient.
    6. Check whether the absolute difference between the estimated eigenvalue and the exact eigenvalue is less than or equal to the specified accuracy.
    7. If the required accuracy is achieved, stop the iteration.
    8. Otherwise, update the current vector with the new normalized vector and repeat the process.
    9. Stop the process if the number of iterations reaches 10,000 to prevent an infinite loop.
    10. Return the final number of iterations required for convergence.

    """
    n = len(matrix)
    v = [1.0] * n
    
    norm_v = vector_norm(v)
    v = [x / norm_v for x in v]
    
    iterations = 0
    while True:
        iterations += 1
        
        v_next = mat_vec_mult(matrix, v)
        norm_v_next = vector_norm(v_next)
        v_next = [x / norm_v_next for x in v_next]
        
        M_v_next = mat_vec_mult(matrix, v_next)
        eigenvalue_approx = vec_dot(v_next, M_v_next)
        
        if abs(eigenvalue_approx - true_val) <= target_acc:
            break
        v = v_next
        
        if iterations > 10000:
            break # Safety limit
            
    return iterations

# 3(e) Compare iterations required for 10^-7 accuracy
iters = count_iterations_for_accuracy(C, actual_vals_np[0], target_acc=1e-7)
print(f"Itations required to approximate λ1 to 10^-7 accuracy: {iters}")
```
